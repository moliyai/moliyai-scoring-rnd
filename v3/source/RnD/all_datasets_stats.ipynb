```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
datasets = {
    "df3m10dif": pd.read_csv('df_3month_10difference.csv'),
    "df3m15dif": pd.read_csv('df_3month_15difference.csv'),
    "df3m30dif": pd.read_csv('df_3month_30difference.csv'),
    "df3m45dif": pd.read_csv('df_3month_45difference.csv'),
    "df3m60dif": pd.read_csv('df_3month_60difference.csv'),
    "df6m10dif": pd.read_csv('df_6month_10difference.csv'),
    "df6m15dif": pd.read_csv('df_6month_15difference.csv'),
    "df6m30dif": pd.read_csv('df_6month_30difference.csv'),
    "df6m45dif": pd.read_csv('df_6month_45difference.csv'),
    "df6m60dif": pd.read_csv('df_6month_60difference.csv')
}
```

```python
for name, df in datasets.items():
    train_df, test_df = train_test_split(
        df, test_size=0.3, random_state=42, stratify=df['is_bad_client']
    )

    npl = (test_df['is_bad_client'].sum() / len(test_df)) * 100

    print(f"{name} → NPL в тестовой выборке: {npl:.2f}%")
```

```python
for name, df in datasets.items():
    print(f"\n{name}")

    counts_before = df['is_bad_client'].value_counts()
    npl_before = (counts_before.get(1, 0) / len(df)) * 100
    print(f"  ДО разделения:  Хорошие = {counts_before.get(0, 0)},  Плохие = {counts_before.get(1, 0)},  NPL = {npl_before:.2f}%")

    train_df, test_df = train_test_split(
        df, test_size=0.3, random_state=42, stratify=df['is_bad_client']
    )

    counts_after = test_df['is_bad_client'].value_counts()
    npl_after = (counts_after.get(1, 0) / len(test_df)) * 100
    print(f"  В тестовой выборке:  Хорошие = {counts_after.get(0, 0)},  Плохие = {counts_after.get(1, 0)},  NPL = {npl_after:.2f}%")
```
