```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../dataset/toyona_gold_bnpl.csv")
df.head()
```

```python
df.columns = [col.lower() for col in df.columns]
```

```python
df_copy = df.copy()
df_copy.columns
```

```python
df_copy.drop(['id клиента', 'уровень образования', 'наличие авто', 'собственное жилье', 'залог'], axis=1, inplace=True)
```

```python
df_copy.head()
```

```python
df_copy = df_copy.apply(lambda x: x.str.strip() if x.dtype == "object" else x)
```

```python
df_copy.loc[(df_copy['пол'] == 'male') & (df_copy['семейное положение'] == 'Женат/Замужем'), 'семейное положение'] = 'женат'
df_copy.loc[(df_copy['пол'] == 'female') & (df_copy['семейное положение'] == 'Женат/Замужем'), 'семейное положение'] = 'замужем'

df_copy.loc[(df_copy['пол'] == 'male') & (df_copy['семейное положение'] == 'Холост/Не замужем'), 'семейное положение'] = 'холост'
df_copy.loc[(df_copy['пол'] == 'female') & (df_copy['семейное положение'] == 'Холост/Не замужем'), 'семейное положение'] = 'не замужем'
```

```python
for col in df_copy.select_dtypes(include=['object']).columns:
    df_copy[col] = df_copy[col].astype(str).str.strip().str.lower()
```

```python
df_copy['профессия'] = df_copy['профессия'].str.strip().str.split().str.join(' ')
```

```python
def replace_product_name(name: str):
    if not isinstance(name, str) or name == "":
        return "кольцо"
    
    first = name.strip()[0].lower()
    if first == "а" or first == "a":
        return "кулон"
    elif first == "к" or first == "k":
        return "комплект"
    elif first == "м" or first == "m" or first == "m":
        return "цепочка"
    elif first == "e":
        return "серьги"
    elif first == "r":
        return "кольцо"
    elif first.isalpha():
        return name.strip()
    else:
        return "кольцо"
```

```python
df_copy["название продукта"] = df_copy["название продукта"].apply(replace_product_name)
```

```python
df_copy['количество иждивенцев'] = pd.to_numeric(df_copy['количество иждивенцев'], errors='coerce')
df_copy['зарплата'] = pd.to_numeric(df_copy['зарплата'], errors='coerce')


df_copy['дата рождения'] = pd.to_datetime(df_copy['дата рождения'], errors='coerce')
df_copy['дата выдачи'] = pd.to_datetime(df_copy['дата выдачи'], errors='coerce')
df_copy['дата окончания'] = pd.to_datetime(df_copy['дата окончания'], errors='coerce')
```
