```python
def replace_product_name(name: str):
    if not isinstance(name, str) or name.strip() == "":
        return "кольцо"

    parts = [part.strip() for part in name.split(";") if part.strip()]

    def map_single(n: str):
        first = n[0].lower()
        if first in ["а", "a"]:
            return "кулон"
        elif first in ["к", "k"]:
            return "комплект"
        elif first in ["м", "m"]:
            return "цепочка"
        elif first == "e":
            return "серьги"
        elif first == "r":
            return "кольцо"
        elif first.isalpha():
            return n
        else:
            return "кольцо"
        
    return "; ".join(map(map_single, parts))
```

```python
df_copy["название продукта"] = df_copy["название продукта"].apply(replace_product_name)
```
